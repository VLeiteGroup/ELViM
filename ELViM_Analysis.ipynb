```python
##Kernel Density Estimation (KDE) with Scatter##

##The KDE can also be calculated for a specific set of points instead of the whole projection##

import numpy as np  # Import NumPy for numerical operations
import matplotlib.pyplot as plt  # Import Matplotlib for plotting
from scipy import stats  # Import statistical functions from SciPy

# Load the projection file
proj = 'proj.out'  # Save the name of the projection file as a string
x, y = np.genfromtxt(proj, unpack='True')  # Read data from the projection file and unpack it into x and y coordinates
xy = np.vstack([x, y])  # Stack the x and y arrays vertically to form a 2D array (shape: [2, N])

# Calculate the standard deviation of the data points
std_dev = np.std(xy, axis=1).mean()  # Compute the standard deviation of the x and y values along axis 1 (rows), then take the average of both standard deviations

# Calculate dynamic bandwidth using a scaling factor (you can adjust the factor as needed)
scaling_factor = 0.5  # This scaling factor controls the dynamic bandwidth size, you can experiment with it (values from 0.05 to 1.0)
bw_method = scaling_factor * std_dev  # Multiply the average standard deviation by the scaling factor to determine the bandwidth

# Estimate the probability density function using the dynamically computed bandwidth
z = stats.gaussian_kde(xy, bw_method=bw_method)(xy)  # Compute the Kernel Density Estimate (KDE) for the points with the calculated bandwidth

# Sort the density values to visualize the points with lower or higher density
idx = np.argsort(z)  # Create an index array that sorts the density values in ascending order

# Create the figure and a set of subplots
fig, ax = plt.subplots()  # Create a figure and a set of axes (subplots) for plotting
plt.gca().set_aspect('equal')  # Ensure the x and y axes are scaled equally (aspect ratio is 1:1)

# Create the scatter plot where color corresponds to the log of the density
sc = ax.scatter(x[idx], y[idx], s=3.0, marker='o', c=(z[idx]), edgecolor='k', linewidth=0.02, alpha=1, cmap='afmhot_r')  
# Plot the x, y data points using a scatter plot. Color points based on the KDE density, with a color map ('afmhot_r')

# Remove the x and y axes
plt.axis('off')  # Disable the axis to give a clean plot without axis labels or ticks

# Create and format the colorbar
cbar = fig.colorbar(sc, ax=ax, pad=0.05, shrink=1)  # Add a colorbar to the plot, with some padding between plot and colorbar
cbar.set_label('Density', fontsize=16)  # Label the colorbar as 'Density', with font size 16
cbar.ax.tick_params(labelsize=16)  # Set the font size for the ticks on the colorbar

# Save the figure as a PNG file
plt.savefig('Density_ELViM.png', format='png', dpi=600, bbox_inches='tight', pad_inches=0.05)  # Save the plot as a high-resolution PNG file with tight bounding box and padding
plt.show()  # Display the plot on the screen
```

```python
##Entropy Estimation##

# Import necessary libraries
import numpy as np              # NumPy for handling arrays and mathematical operations
import matplotlib.pyplot as plt  # Matplotlib for plotting (though not used explicitly here)
from scipy import stats         # SciPy for statistical methods, including 2D binned statistics

# Define the input projection file name
proj = 'proj.out'  ##Save the name of the projection file##

# Load the X and Y coordinates from the projection file into separate arrays
x, y, = np.genfromtxt(proj, unpack='True')  ##Save the coordinates X and Y from the projection file##

# Assign uniform weights (1) to each data point in the dataset
w = np.ones(len(x))  ##Assigns the same weight to all points##

# Calculate the 2D binned statistic (counts of points in each 2D bin) over the range of the data
# `bins=[25, 25]` means the data will be divided into a 25x25 grid
# `expand_binnumbers=True` allows for expanding the binning numbers for later use
stat, x_edge, y_edge, binnumber = stats.binned_statistic_2d(x, y, w, 'count', bins=[25, 25], expand_binnumbers=True)
##

############  countagem de amostras por bin para tipo 1
# Re-calculate the binned statistic with the same weights but this time using the edges from the previous calculation
w = np.ones(len(x))  ##Assigns the same weight to all points##

# Use the edges of the bins from the previous `binned_statistic_2d` result
# The `x_edge` and `y_edge` represent the bin edges, so this uses the exact same binning as before
stat1, _, _, binn1 = stats.binned_statistic_2d(x, y, w, 'count', bins=[x_edge, y_edge], expand_binnumbers=True)
##Set the grid and count the number of points in each bin##

# Normalize the 2D histogram by dividing by the sum of all counts
# This turns the counts into a probability distribution (i.e., the sum of all values in `stat1` will equal 1)
stat1 = stat1 / np.sum(stat1)  ##Normalizes the function##

# Replace zero values with 1 to avoid taking the logarithm of zero, which is undefined
# This ensures the entropy calculation proceeds without encountering NaNs or infinities
stat1[stat1 == 0] = 1  ##Assigns all zeros equal 1 to compute the log##

# Compute the entropy using the formula:
# H = -Σ (P(x) * log(P(x))) where P(x) is the probability distribution in `stat1`
H = -np.sum(stat1 * np.log(stat1))  ##Compute the log of the distribution##

# Print the computed entropy value
print(H)  ##Return the value of the entropy on the screen##
```

```python
##Comparison between different KDE##

##The KDE can be calculated for a specific set of points instead of the whole projection##
##With this analysis, one can compare two different sets of KDE that came from the same projection##

# Import necessary libraries
import numpy as np              # NumPy for handling arrays and mathematical operations
import matplotlib.pyplot as plt  # Matplotlib for plotting
from scipy import stats         # SciPy for statistical methods, specifically Kernel Density Estimation (KDE)
import seaborn as sns; sns.set(style="white", color_codes=True)  # Seaborn for setting plotting style (white background and color codes)

# Load data from 'proj_all.dat' (whole projection data)
xall, yall = np.genfromtxt('proj.out', unpack=True)  ##Loads the whole projection to define the single grid##

# Load data for the first and second sets of points (e.g., different types of tetramers)
x1, y1 = np.genfromtxt('ab40_tetramers.dat', unpack=True)  ##Loads the set of the first KDE##
x2, y2 = np.genfromtxt('ab42_tetramers.dat', unpack=True)  ##Loads the set of the second KDE##

# Concatenate the whole projection data into a 2D array
values = np.vstack([xall, yall])  ##Concatenate the whole projection##

# Define the grid boundaries with a small padding of 0.05 to avoid edge effects
xmax = xall.max() + 0.05          # Find the maximum x value and add 0.05 to avoid edge effects
xmin = xall.min() - 0.05          # Find the minimum x value and subtract 0.05
ymax = yall.max() + 0.05          # Find the maximum y value and add 0.05 to avoid edge effects
ymin = yall.min() - 0.05          # Find the minimum y value and subtract 0.05

# Create a 2D mesh grid for the KDE calculation with 300 bins along both axes
X, Y = np.mgrid[xmin:xmax:300j, ymin:ymax:300j]  ##Set the grid with (300 x 300) bins

# Flatten the X and Y grid arrays and concatenate them into a single array of positions
positions = np.vstack([X.ravel(), Y.ravel()])   ##Concatenate the X and Y positions in a scipy-compatible format##

# Prepare the data for the first set of points
values1 = np.vstack([x1, y1])  # Stack the x1 and y1 values into a 2D array

# Compute the Kernel Density Estimation (KDE) for the first set of points
kernel1 = stats.gaussian_kde(values1)  ##Computes the KDE for the first set##
kernel1.set_bandwidth(bw_method=kernel1.factor * 0.4)  # Set the bandwidth for KDE (adjust based on the factor)
c1 = kernel1.evaluate(values)  # Evaluate the KDE at the locations defined by 'values' (entire data set)
Z1 = np.reshape(kernel1(positions).T, X.shape)  # Reshape the KDE results to match the shape of the mesh grid

# Prepare the data for the second set of points
values2 = np.vstack([x2, y2])  # Stack the x2 and y2 values into a 2D array

# Compute the Kernel Density Estimation (KDE) for the second set of points
kernel2 = stats.gaussian_kde(values2)  # Compute the KDE for the second set of points
kernel2.set_bandwidth(bw_method=kernel2.factor * 0.4)  ##Computes the KDE for the second set##
c2 = kernel2.evaluate(values)  # Evaluate the KDE for the second set of points at the same locations as the first set
Z2 = np.reshape(kernel2(positions).T, X.shape)  # Reshape the KDE results to match the shape of the mesh grid

# Compute the difference between the two KDEs (first set minus second set)
Z = Z1 - Z2  ##Computes the difference between the two KDEs##

# Define the color scale limits to make the plot symmetric
lim = max(abs(np.max(Z)), abs(np.min(Z)))  ##Applies a limit to make the scale symmetrical##
# This ensures that the color range of the plot is centered around zero, maintaining a balanced view of the KDE differences

# Create a new figure for the plot
plt.figure()  ##Creates the figure##

# Scatter plot the entire projection data for reference (in black with very low opacity)
plt.scatter(xall, yall, s=1.0, marker='o', c='k', linewidth=0, alpha=0.05)  ##Plot the points of both sets for reference##
# This allows you to see where the points are located in the overall space, but with a very low opacity so they don't overpower the contours

# Plot the difference in KDEs as an image with the calculated limits and the RdBu colormap
img = plt.imshow(Z.T, origin='lower', extent=[xmin, xmax, ymin, ymax], cmap="RdBu")  ##Plot the KDE differences##
# The `imshow` function displays the KDE difference array as an image. The `RdBu` colormap helps highlight positive and negative differences.

# Set the color scale limits (from -lim to +lim) to ensure zero is the center of the color scale
plt.clim(-lim, lim)  ##Defines the zero as the center of the scale##

# Add a colorbar to the plot to show the scale of the KDE differences
plt.colorbar(img)  ##Creates a colorbar##
# The colorbar provides a reference for the values represented by the colors in the plot.

# Remove the x and y axis labels and ticks for a cleaner plot
plt.axis('off')  ##Remove the X and Y axis##

# Display the final plot
plt.show()  ##Plot the whole figure##
```

```python
##Free energy estimation##

# Import necessary libraries
import numpy as np             # NumPy for handling arrays and mathematical operations
import matplotlib.pyplot as plt # Matplotlib for plotting
import math                     # Math module for mathematical functions (though not directly used here)
import matplotlib as mpl        # For advanced colormap handling and color manipulation
import mdtraj as md            # MDTraj library (though not used here, it's typically for molecular dynamics analysis)
from scipy import stats         # SciPy for statistical methods, including Kernel Density Estimation (KDE)

# Load data from 'proj.out' file
x, y = np.genfromtxt('proj.out', unpack=True)  # Load x and y coordinates from a file, 'proj.out', unpacking into two separate arrays

# Determine the min and max values for x and y coordinates
xmax = x.max()  # Find the maximum value in the x array
xmin = x.min()  # Find the minimum value in the x array
ymax = y.max()  # Find the maximum value in the y array
ymin = y.min()  # Find the minimum value in the y array

# Define a 2D grid (mesh) for the KDE calculation, with 300x300 bins
X, Y = np.mgrid[xmin:xmax:300j, ymin:ymax:300j]  # Create a mesh grid between the min and max values of x and y, with 300 divisions (bins)

# Prepare the data for KDE calculation
positions = np.vstack([X.ravel(), Y.ravel()])  # Flatten X and Y mesh grids and stack them into a 2D array
values = np.vstack([x, y])                    # Stack the original x and y data into a 2D array

# Perform Kernel Density Estimation (KDE) to estimate the probability density function
kernel = stats.gaussian_kde(values)  # Create a KDE object using the x and y data points
kernel.set_bandwidth(bw_method=kernel.factor*0.5)  # Set the bandwidth of the KDE using a factor based on the kernel's factor (0.5 here)

# Compute the KDE for each position in the grid
Z = np.reshape(kernel(positions).T, X.shape)  # Apply the kernel to the grid positions and reshape the result into the shape of X/Y grid

# Calculate the free energy (FE) from the KDE
FE = -np.log(Z)  # Take the negative log of the KDE values to estimate free energy (in units of kBT)
FE = FE - np.min(FE)  # Normalize the free energy by subtracting the minimum value

# Define the color range and levels for the contour plot
min_val, max_val = 0.1, 1.0  # Define the minimum and maximum values for the color range
n = 10  # Number of colors in the colormap
orig_cmap = plt.cm.Purples  # Use the 'Purples' colormap from Matplotlib
colors = orig_cmap(np.linspace(min_val, max_val, n))  # Generate a set of colors based on the colormap and range
colors = colors[::-1]  # Reverse the color order for the colormap
cmap = mpl.colors.LinearSegmentedColormap.from_list("mycmap", colors)  # Create a new colormap from the reversed colors

# Set contour levels and plot
levels = np.arange(0, 4.5, 0.5)  # Define the contour levels for the plot (0, 0.5, 1.0, ..., 4.0)

# Create a figure and axis for plotting
fig, ax = plt.subplots()  # Create a new figure and axis object
plt.gca().set_aspect('equal')  # Set the aspect ratio of the plot to be equal (squared)

# Plot filled contour plot for free energy
cs = ax.contourf(X, Y, FE, cmap="YlGnBu_r", extend='max', levels=levels, extent=[xmin, xmax, ymin, ymax], alpha=1)
# Create filled contour plot using the 'YlGnBu_r' colormap, with specified levels and plot range (extent)

# Overlay contour lines on top of the filled contour plot
ax.contour(X, Y, FE, colors='darkslateblue', extend='max', levels=levels, linewidths=1, extent=[xmin, xmax, ymin, ymax], alpha=0.7)
# Plot contour lines on top with specified color, linewidth, and transparency (alpha)

# Optionally, scatter original data points (currently commented out)
# ax.scatter(x, y, s=1.0, marker='o', c='k', linewidth=0.0, edgecolor='k', alpha=0.2, zorder=1)
# This line, if uncommented, would add scattered points from x, y on the plot

# Hide axes (no x and y ticks/labels on the plot)
plt.axis('off')  # Turn off the axis labels and ticks

# Add a colorbar to the plot, with specified ticks and label
cbar = fig.colorbar(cs, ax=ax, ticks=[0, 1.0, 2.0, 3.0, 4.0], pad=0.05, shrink=0.8)  # Add a colorbar with specific ticks
cbar.set_label(r'Free Energy/$k_\beta T$', fontsize=18)  # Set the label for the colorbar
ax.tick_params(width=1.5)  # Set the width of the ticks on the plot axes

# Set the x and y axis limits to match the input data's min and max
ax.set_xlim(xmin, xmax)  # Set x-axis limits based on input data
ax.set_ylim(ymin, ymax)  # Set y-axis limits based on input data

# Disable autoscaling for the axes (fix the limits)
ax.autoscale(False)  # Do not automatically adjust the axes limits

# Adjust layout to prevent overlapping of plot elements
plt.tight_layout()  # Automatically adjust the padding of the plot elements

# Save the plot as a high-resolution PNG image
plt.savefig('FE_ELViM.png', format='png', dpi=1200, bbox_inches='tight', pad_inches=0.1)  # Save the plot to a file with high resolution

# Display the plot
plt.show()  # Show the plot in a window
```
